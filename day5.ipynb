```python
class Student:
  def __init__(self,name,roll_no):
    print("auto start")
    self.name = name
    self.roll_no = roll_no
  # get
  def updateName(self,name=None): # default value -> none
    if name != None:
      self.name = name
    print(self.name)


inst = Student("nikhil",123)
inst.updateName("rajendra")

inst1 = Student("abhishek",321)
inst1.updateName()
```
