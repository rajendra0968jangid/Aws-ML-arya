```python
# 1. Supervised Learning
# Meaning

# In supervised learning:

# Input data available
# Correct output also available

# Model learns from examples.
```

```python
# Types of Supervised Learning
# A) Regression

# Used when output is a number.

# Examples:

# House price prediction
# Temperature prediction
# Salary prediction

# B) Classification

# Used when output is category/class.

# Examples:

# Spam / Not Spam
# Pass / Fail
# Disease / No Disease
```

```python
# 2. Unsupervised Learning
# Meaning

# Only input data available.
# No output labels.

# Model finds hidden patterns itself.
```

```python
# Real Life Example

# Suppose shop owner has customer data:

# Age
# Salary
# Shopping habits

# Model automatically creates groups:

# Rich customers
# Medium customers
# Low budget customers

# This is clustering.
```

```python
# Types of Unsupervised Learning
# A) Clustering

# Grouping similar data.

# Examples:

# Customer segmentation
# Friend suggestions
```

```python
# B) Dimensionality Reduction

# Reducing unnecessary columns/features.

# Example:

# 100 columns → reduce to 10 important columns

# Benefits:

# Faster training
# Less memory
# Better performance
```

```python
# C) Association Rule Learning

# Finding item relationships.

# Example:

# People buying bread also buy butter.

# Used in:

# Amazon
# BigBasket
# Flipkart
```

```python
# 3. Semi-Supervised Learning
# Meaning

# Small labeled data + large unlabeled data.
```

```python
# Real Life Example

# Suppose:

# 1000 cat images
# Only 100 labeled

# Model learns from both labeled and unlabeled images.

# Used when labeling is expensive.
```

```python
# 4. Reinforcement Learning
# Meaning

# Model learns using:

# Reward
# Punishment
```

```python
# Real Life Example
# Teaching Dog
# Correct action → reward
# Wrong action → punishment

# Similarly:
# AI learns best actions.
```

**4 DATA PREPROCESSING**
1. Before training ML model: data must be cleaned.

```python
# 4.1 Handling Missing Values
# Meaning

# Dataset may contain missing data.

# Example:

# Name	Salary
# Ram	  20000
# Shyam NaN

# NaN = missing value.
```

```python
# Solutions
# 1. Remove rows

# If few missing values.

# 2. Fill values

# Using:

# Mean
# Median
# Mode
```

```python
# 4.2 Handling Outliers
# Meaning

# Very abnormal values.

# Example salaries:

# 20k, 25k, 30k, 5000000

# 5000000 is outlier.
```

```python
# Problems

# Outliers can:

# disturb model
# reduce accuracy
```

```python
# 4.2.1 IQR Method
# Meaning

# Finds values too far from normal range.

# Good for:

# skewed data (ratio-> 90:20)
```

```python
# Z-Score Method

# Measures:

# How far value is from mean.

# High z-score = possible outlier.
```

```python
# 4.3 Encoding Categorical Data

# ML model works with numbers.

# Text must convert into numeric form.
```

```python
# 4.3.1 Label Encoding

# Converts labels into numbers.
```

```python
# 4.3.1 One-Hot Encoding

# Creates separate columns.
```

```python
# 4.4 Feature Scaling
# Meaning

# Bring all features into same scale.
```

```python
# 4.4.1 Normalization (Min-Max Scaling)

# Converts values between:
# 0 to 1
```

```python
# 4.4.2 Standardization (Z-Score)

# Transforms data:

# Mean = 0
# Std = 1
```

```python
# 4.5 Handling Class Imbalance
# Problem

# Example:

# Class	Count
# Normal	950
# Fraud	50

# Model may always predict:

# Normal

# Still gets high accuracy.
```

```python
# MODEL PROBLEMS
# 4.6 Overfitting vs Underfitting
```

```python
# Underfitting
# Meaning

# Model learns too little.

# Poor performance on:

# training data
# testing data

# Example

# Student studied only 1 page for exam.

# Result:

# fails everywhere
```

```python
# Overfitting
# Meaning

# Model memorizes training data.

# Good on training data.
# Bad on new data.

# Example

# Student memorized exact answers only.

# New question comes:

# fails
```

**Regression Metrics**

```python
# MAE (Mean Absolute Error)

# Average absolute error.

# Example:
# Prediction error:

# 2, 3, 5

# Average:

# 3.3
```

```python
# RMSE

# Square root of MSE.

# Most interpretable.
```
